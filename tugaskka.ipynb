````python
#latihan 4

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di
        # dalam class ini)
        # Initialize hp using the setter for validation
        self.__hp = 0 # Temporary initialisation before calling set_hp
        self.set_hp(hp_awal)

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar
        # aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
print("\n--- Demonstrasi Enkapsulasi --- ")
hero1 = Hero("Layla", 999999) # Coba set HP sangat tinggi
print(f"Initial HP for {hero1.nama}: {hero1.get_hp()}")

hero2 = Hero("Miya", 100) # Normal HP
print(f"Initial HP for {hero2.nama}: {hero2.get_hp()}")

hero1.set_hp(-50) # Coba set negatif
print(f"HP for {hero1.nama} after set_hp(-50): {hero1.get_hp()}") # Output: 0 (Karena dicegat oleh logika Setter)

print("\n--- Demonstrasi Serangan --- ")
hero1.diserang(20) # Layla diserang 20
hero2.diserang(hero1.get_hp() + 50) # Miya menyerang Layla dengan damage besar
print(f"HP for {hero1.nama} after attacks: {hero1.get_hp()}")

hero2.diserang(50) # Miya diserang 50
print(f"HP for {hero2.nama} after attack: {hero2.get_hp()}")
````

````python
from abc import ABC, abstractmethod
# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
````

1.Jika abstract method tidak diimplementasikan: Class turunan tidak bisa dibuat objek karena masih dianggap 'abstrak'. Pesan errornya adalah TypeError yang menyatakan kita tidak bisa 'instantiate' class abstrak. Ini memastikan bahwa setiap class yang menuruni 'kontrak' abstrak harus menyediakan semua method yang dijanjikan.

2.Mengapa Abstract Class seperti GameUnit tidak bisa dibuat objek langsung: Class abstrak berfungsi sebagai cetak biru atau 'kontrak' yang mendefinisikan perilaku umum (melalui abstract method) yang harus diimplementasikan oleh class turunannya. Ia penting untuk:
Konsistensi: Memastikan semua turunan memiliki method yang sama.
Polimorfisme: Memungkinkan objek turunan diperlakukan sebagai tipe dasar yang sama.
Validasi: Mencegah pembuatan objek dari class yang belum lengkap implementasinya

````python
#latihan 6


# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama
    def serang(self):
        print("Hero menyerang dengan tangan kosong.")
# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")
# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")
# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")
# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]
print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
````
